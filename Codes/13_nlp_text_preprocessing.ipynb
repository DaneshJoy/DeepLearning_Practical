```python
from keras.layers import TextVectorization
```

## Create a Corpus

```python
corpus = ["I like my brother.",
          "Go to Home!",
          "Do you like my brother?"]
```

## Adapt a vectorizer on our corpus

```python
text_vectorizer = TextVectorization(ngrams=2)
text_vectorizer.adapt(corpus)
```

```python
vocab = text_vectorizer.get_vocabulary()
vocab
```

```python
for i, word in enumerate(vocab):
    print(f'{word:<9}: {i}')
```

## Create sequence vectors of our corpus

```python
encoded_sequences = text_vectorizer(corpus)
print(encoded_sequences)
```

```python
encoded_sequences.numpy()
```

## Test on new sequence

```python
encoded_test = text_vectorizer(['I like my sister'])
print(encoded_test)
```
